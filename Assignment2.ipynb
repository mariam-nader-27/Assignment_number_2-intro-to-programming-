```python
def greater_than_3(words, K):
    new_list=[]
    for word in words:
        if len(word)>K:
            new_list.append(word)
    return new_list       
words = ["apple", "pie", "banana"]
K = 3

print(greater_than_3(words, K))
```

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []
        
    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price
        
    def book_tables(self, customer_name, table_number):
        self.book_table.append({"customer": customer_name,"table": table_number})
        
    def customer_order(self, customer_name, items):
        self.customer_orders.append({ "customer": customer_name, "items": items})
    
    def print_menu(self):
        print("Menu:")
        for item, price in self.menu_items.items():
            print(item, ":", price)

    def print_table_reservations(self):
        print("Table Reservations:")
        for r in self.book_table:
            print(r)

    def print_customer_orders(self):
        print("Customer Orders:")
        for order in self.customer_orders:
            print(order)

restaurant = Restaurant()

restaurant.add_item_to_menu("Cake", 150)
restaurant.add_item_to_menu("Coffee", 100)
restaurant.add_item_to_menu("Pizza", 120)
restaurant.book_tables("Mariam", 1)
restaurant.book_tables("Malak", 2)
restaurant.customer_order("Mariam", ["Cake", "Coffee"])
restaurant.customer_order("Malak", ["Pizza"])
restaurant.print_menu()
print("___________________________________\n")
restaurant.print_table_reservations()
print("___________________________________\n")
restaurant.print_customer_orders()
```

```python
class Customer:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    def get_name(self):
        return self._name
    
    def get_age(self):
        return self._age
    
    def set_name(self, name):
        self._name = name

    def set_age(self, age):
        self._age = age

class MainBikeRental:
    def __init__(self):
        self._bikes = []
        self._rentals = []

    def get_bikes(self):
        return self._bikes

    def get_rentals(self):
        return self._rentals

    def set_bikes(self, bikes):
        self._bikes = bikes
    
    def set_rentals(self, rentals):
        self._rentals = rentals

    

    def requestBike(self, customer, number_of_bikes):
        if customer.get_age() <= 6:
            print("sorry you must be over 6 years old.")
            return

        if number_of_bikes <= 0:
            print("Error.")
            return

        if number_of_bikes <= len(self._bikes):
            rented_bikes = self._bikes[:number_of_bikes]
            del self._bikes[:number_of_bikes]
            self._rentals.append({"customer": customer,"bikes": rented_bikes})
            print("Bikes rented successfully.")
            
        else:
            print("Wait 10 mins")

    def returnBike(self, rentalTime, num_of_bikes):
        if not self._rentals:
            print("No active rentals.")
            return

        if rentalTime < 0 or num_of_bikes <= 0:
            print("error,try agian.")
            return

        rental = self._rentals[-1]

        if num_of_bikes > len(rental["bikes"]):
            print("Invalid number of returned bikes.")
            return

        returned_bikes = rental["bikes"][:num_of_bikes]
        self._bikes.extend(returned_bikes)
        del rental["bikes"][:num_of_bikes]

        if len(rental["bikes"]) == 0:
            self._rentals.pop()

        total_bill = rentalTime * 40
        print("Total Bill:", total_bill)
    
    def addBike(self, bike):
            self._bikes.append(bike)

    def totalCost(self):
        total = 0

        for bike in self._bikes:
            total += bike.get_price()

        return total

class Bike:
    def __init__(self, name, price):
        self._name = name
        self._price = price

    def get_name(self):
        return self._name

    def set_name(self, name):
        self._name = name

    def get_price(self):
        return self._price

    def set_price(self, price):
        self._price = price

customer1 = Customer("Ali", 18)

bike1 = Bike("streat125 Bike", 378)
bike2 = Bike("streat26 Bike", 209)
bike3 = Bike("streat3 Bike", 350)

shop = MainBikeRental()
shop.addBike(bike1)
shop.addBike(bike2)
shop.addBike(bike3)

print("Initial Stock Value:", shop.totalCost())

shop.requestBike(customer1, 2)
print("___________________________________\n")
print("Stock Value After Rental:", shop.totalCost())
shop.returnBike(3, 1)
print("___________________________________\n")
print("Stock Value After Return:", shop.totalCost())

```
